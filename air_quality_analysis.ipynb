# Air Quality Data Analysis using Python

This project performs exploratory data analysis on air pollution data from Indian cities.
The goal is to analyze pollution trends, compare cities, and visualize pollution patterns.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python


df = pd.read_csv("data/city_day.csv")
```

```python
df['Date'] = pd.to_datetime(df['Date'])
```

```python
df['PM2.5'].mean()
```

```python
city_pollution = df.groupby('City')['PM2.5'].mean().sort_values(ascending=False)
```

```python
plt.figure(figsize=(12,6))

city_pollution.head(10).plot(kind='bar')

plt.title("Top 10 Most Polluted Cities (PM2.5)")
plt.ylabel("Average PM2.5")
plt.xlabel("City")

plt.tight_layout()

plt.savefig("images/top_polluted_cities.png")

plt.show()
```

### PM2.5 Pollution Trend

This graph shows how PM2.5 pollution levels change over time.
Higher values indicate worse air quality in cities.

```python
plt.savefig("images/top_polluted_cities.png")
```

```python
df['Year'] = df['Date'].dt.year
```

```python
pm25_yearly = df.groupby('Year')['PM2.5'].mean()
pm25_yearly
```

```python
pm25_yearly.plot(kind='line', marker='o', figsize=(8,5))
plt.title("Average PM2.5 Trend Over Years")
plt.ylabel("PM2.5")
plt.xlabel("Year")
plt.show()
```

### PM2.5 Trend Over Years

This graph shows how the average PM2.5 pollution level changes over different years.  
The trend helps identify whether air quality is improving or worsening over time.  
Changes in the trend may reflect environmental policies, industrial activity, or seasonal variations.

```python
plt.savefig("images/pm25_yearly_trend.png")
```

```python
import seaborn as sns

pollution_cols = ['PM2.5','PM10','NO','NO2','NOx','NH3','CO','SO2','O3']
corr = df[pollution_cols].corr()
```

```python
plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap="coolwarm")
plt.title("Correlation Between Pollutants")
plt.show()
```

### Correlation Between Pollutants

The heatmap shows relationships between different air pollutants such as PM2.5, PM10, NO2, and SO2.  
Higher correlation values indicate pollutants that tend to increase or decrease together.  
Understanding these relationships helps identify possible common pollution sources.

```python
plt.savefig("images/pollution_correlation.png")
```

```python
plt.figure(figsize=(8,5))
sns.histplot(df['PM2.5'], bins=50)
plt.title("Distribution of PM2.5 Pollution")
plt.show()
```

### Distribution of PM2.5 Pollution

This histogram shows how frequently different PM2.5 values occur in the dataset.  
It helps understand the overall spread and concentration of pollution levels across cities.  
Higher frequencies in certain ranges indicate common pollution levels experienced in the data.

```python
plt.savefig("images/pm25_distribution.png")
```
